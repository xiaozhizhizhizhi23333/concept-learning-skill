### 练习 4 参考写法
```python

weight = float(input("请输入体重(公斤)："))
bmi = weight / (height ** 2)
print(f"你的BMI是 {bmi:.1f}")
```

### 练习 5 参考写法
```python
if bmi < 18.5:
    print("偏瘦")
elif bmi < 24:
    print("正常")
else:
    print("偏胖")
```

**注意**：这里的判断顺序很关键——`elif bmi < 24` 之所以不用写「并且大于等于 18.5」，是因为程序从上往下判断，能走到这一行就说明前面的条件已经不成立了。这是个小技巧，体会一下。
